```python
empleados_buenos = empleados[empleados['Attrition']=='No']
empleados_malos = empleados[empleados['Attrition']=='Yes']
empleados_buenos , empleados_malos
```

```python
plt.scatter(empleados_buenos['DistanceFromHome'], empleados_buenos['YearsAtCompany'], color='green', label='no renuncia',marker='d', s=90)
plt.scatter(empleados_malos['DistanceFromHome'], empleados_malos['YearsAtCompany'], color='red', label='renuncia',marker='o',s=90)
plt.title('Distancia desde casa vs años en la empresa')
plt.xlabel('Distancia desde casa')
plt.ylabel('Años en la empresa')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
#podemos notar que los empleados que renuncian no precisamente viven cerca ,por lo que la causa de sus renuncuas no se debe a la distancia  .
```

```python
print(empleados_malos[['Age','YearsAtCompany']].mean())
print(empleados_buenos[['Age','YearsAtCompany']].mean())
#supongamos que mayor edad significa mayor experiencia y por tanto mayor salario,y que mas años en la empresa significa mayor salario,  .

```

```python
plt.scatter(empleados_buenos['MonthlyIncome'], empleados_buenos['YearsAtCompany'], color='green', label='no renuncia',marker='d', s=90)
plt.scatter(empleados_malos['MonthlyIncome'], empleados_malos['YearsAtCompany'], color='red', label='renuncia',marker='o',s=90)
plt.title('Salario mensual vs años en la empresa')
plt.xlabel('Salario mensual')
plt.ylabel('Años en la empresa')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
#particularmente podemos notar que los empleados que renuncian tienen un salario mensual menor que los que no renuncian, por lo que podemos suponer que la causa de sus renuncias se debe a un salario bajo,y que la distancia desde casa no es un factor determinante para que un empleado renuncie.
```

```python
#despues de aver explicado la base de datos y tratar de explicar las renuncias de los empleados ,usaremos KNN para predecir si un empleado renunciara o no .
from sklearn import preprocessing


datos =empleados[['YearsAtCompany','Age','MonthlyIncome']]
clase=empleados['Attrition']
escalador = preprocessing.MinMaxScaler()
datos = escalador.fit_transform(datos)
datos
#escalamos los datos para que esten entre 0 y 1, ya que KNN es sensible a la escala de los datos.
```

```python

from sklearn.neighbors import KNeighborsClassifier

clasificador = KNeighborsClassifier(n_neighbors=5)
clasificador.fit(datos, clase)
#creamos un clasificador KNN con 5 vecinos y lo entrenamos con los datos escalados y la clase (si el empleado renuncia o no).
```
